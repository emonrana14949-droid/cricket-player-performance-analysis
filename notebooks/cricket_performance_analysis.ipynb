# Cricket Player Performance Analysis

A Pandas, NumPy and Matplotlib walkthrough of the synthetic cricket batting dataset. Player names are real Bangladesh cricketers, but all numbers are made up for practice.


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

plt.style.use('seaborn-v0_8-whitegrid')
df = pd.read_csv('../data/cricket_performance_dataset.csv')
df.head()
```

## 1. Player summary — batting average, strike rate, 50s and 100s


```python
df['times_out'] = np.where(df['dismissal'] == 'Out', 1, 0)
df['is_fifty'] = np.where((df['runs'] >= 50) & (df['runs'] < 100), 1, 0)
df['is_hundred'] = np.where(df['runs'] >= 100, 1, 0)

summary = df.groupby('player').agg(
    innings=('match_id', 'count'),
    total_runs=('runs', 'sum'),
    times_out=('times_out', 'sum'),
    avg_strike_rate=('strike_rate', 'mean'),
    fifties=('is_fifty', 'sum'),
    hundreds=('is_hundred', 'sum'),
    highest_score=('runs', 'max')
).reset_index()
summary['batting_average'] = summary['total_runs'] / summary['times_out']
summary.round(2).sort_values('batting_average', ascending=False)
```

## 2. Visualization: batting average by player


```python
summary.set_index('player')['batting_average'].sort_values().plot(kind='barh', figsize=(9, 5), color='#2f81f7')
plt.title('Batting Average by Player')
plt.xlabel('Batting average')
plt.ylabel('Player')
plt.show()
```
